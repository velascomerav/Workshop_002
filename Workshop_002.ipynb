```python
import matplotlib.pyplot as plt
import sqlite3
import pandas as pd

con = sqlite3.connect("workshop.db")

dashboard_data = pd.read_sql("""
    SELECT *
    FROM merged_data
""", con)

print("Registros cargados desde la base de datos:", len(dashboard_data))
```

```python
genre_popularity = (
    dashboard_data.groupby("track_genre")["popularity"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

genre_popularity.plot(kind="bar", figsize=(10, 5))
plt.title("Top 10 géneros por popularidad promedio")
plt.xlabel("Género musical")
plt.ylabel("Popularidad promedio")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
grammy_year = (
    dashboard_data[dashboard_data["winner"] == True]
    .groupby("year")
    .size()
)

grammy_year.plot(kind="line", figsize=(10, 5), marker="o")
plt.title("Registros Grammy ganadores por año")
plt.xlabel("Año")
plt.ylabel("Cantidad de registros")
plt.grid()
plt.tight_layout()
plt.show()
```
